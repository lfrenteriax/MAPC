# Problemas de Valor en la Frontera (PVF)

## Introducción

Los **Problemas de Valor en la Frontera (PVF)** consisten en encontrar la solución de una ecuación diferencial que satisface condiciones impuestas en los extremos de un intervalo.

A diferencia de los problemas de valor inicial, las condiciones se especifican en dos o más puntos del dominio.

---

## Definición

Un problema de valor en la frontera puede escribirse como:

$$
y'' = f(x,y,y')
$$

con condiciones de frontera:

$$
y(a)=\alpha,\qquad y(b)=\beta.
$$

El objetivo es encontrar una función \(y(x)\) que satisfaga tanto la ecuación diferencial como las condiciones de frontera.

---

## Aplicaciones

Los Problemas de Valor en la Frontera (PVF) tienen aplicaciones en diversas áreas de la ciencia y la ingeniería. Algunos ejemplos son:

- **Transferencia de calor:** Determinar la distribución de temperatura en una barra metálica cuyos extremos se mantienen a temperaturas fijas.

- **Vibraciones de estructuras:** Calcular la deformación de una viga o un puente sometido a cargas, con ambos extremos apoyados o empotrados.

- **Mecánica de fluidos:** Estudiar el perfil de velocidad de un fluido que circula por una tubería con condiciones impuestas en las paredes.

- **Electrostática:** Encontrar el potencial eléctrico en una región donde el potencial es conocido en la frontera, como en un capacitor.

- **Ingeniería civil:** Analizar la deflexión de vigas en edificios y puentes bajo diferentes cargas y restricciones.

- **Ingeniería mecánica:** Diseñar ejes, resortes y componentes mecánicos considerando deformaciones y esfuerzos con condiciones de soporte específicas.

- **Difusión de contaminantes:** Modelar la concentración de una sustancia en un río o en un material, fijando la concentración en los límites del dominio.

- **Biomedicina:** Modelar la distribución de medicamentos o la difusión de oxígeno en tejidos biológicos con condiciones en sus bordes.

---

## Métodos de solución

1. **Método del disparo:** Convierte el problema de valor en la frontera en un problema de valor inicial. Se realiza una estimación de las condiciones iniciales y se ajusta iterativamente hasta satisfacer las condiciones en el extremo final.

2. **Método de diferencias finitas:** Sustituye las derivadas por aproximaciones mediante diferencias entre puntos de una malla, transformando la ecuación diferencial en un sistema de ecuaciones algebraicas.

3. **Método de elementos finitos:** Divide el dominio en pequeños elementos y aproxima la solución mediante funciones definidas en cada uno. Es ampliamente utilizado para resolver problemas con geometrías complejas.

4. **Métodos espectrales:** Aproximan la solución como una combinación de funciones globales (como polinomios o series de Fourier). Ofrecen alta precisión cuando la solución es suave.
